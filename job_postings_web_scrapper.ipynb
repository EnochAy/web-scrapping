```python
import requests
from bs4 import BeautifulSoup

# Define the URL of the job listing page
BASE_URL = "https://www.wemabank.com/careers/job-openings"

def scrape_wema_bank_jobs(keyword="Data Engineer", location="Nigeria"):
    try:
        # Add headers to mimic a browser
        headers = {
            "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36"
        }
        response = requests.get(BASE_URL, headers=headers)
        response.raise_for_status()

        soup = BeautifulSoup(response.text, "html.parser")

        # Adjust tags and classes based on actual page structure
        job_listings = soup.find_all("div", class_="job-posting")  # Replace 'job-posting' with actual class

        relevant_jobs = []
        for job in job_listings:
            title = job.find("h2").text.strip()  # Replace 'h2' with actual tag
            location_info = job.find("span", class_="location").text.strip()  # Replace 'location' with actual class

            if keyword.lower() in title.lower() and location.lower() in location_info.lower():
                job_url = job.find("a", href=True)["href"]
                relevant_jobs.append({"title": title, "location": location_info, "url": job_url})

        if relevant_jobs:
            print(f"Found {len(relevant_jobs)} job(s):")
            for job in relevant_jobs:
                print(f"Title: {job['title']}\nLocation: {job['location']}\nLink: {job['url']}\n")
        else:
            print("No relevant jobs found.")

    except Exception as e:
        print(f"Error: {e}")

# Call the function
scrape_wema_bank_jobs()
```
